## A first calculation

The next cell imports NumPy and Matplotlib, creates a simple varying magnetic-field component, and plots it. Running it successfully confirms that the basic scientific Python packages are available.

````python
import matplotlib.pyplot as plt
import numpy as np

time = np.linspace(0, 30, 500)
bz   = 5 * np.sin(2 * np.pi * time / 10)

fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(time, bz)
ax.axhline(0, color="black", linewidth=0.8)
ax.set(xlabel="Time [min]", ylabel=r"$B_z$ [nT]", title="Example magnetic-field variation")
ax.grid(alpha=0.3)

plt.show()
````
